```python
from Bio import Entrez
```

```python
def find_pubmed_record(doi):
    Entrez.email = "k.aryanpoo@gmail.com"
    try:
        handle = Entrez.esearch(db="pubmed", term=doi, retmax=1)
        search_results = Entrez.read(handle)
        handle.close()

        # Check if any results are found
        if search_results["IdList"]:
            return search_results["IdList"][0]  # Return the first PMID found
        else:
            return "No matching record found"
    except Exception as e:
        return str(e)
```

```python
def get_abstract(pmid):
    Entrez.email = "k.aryanpoo@gmail.com"  # Replace with your email

    try:
        handle = Entrez.efetch(db="pubmed", id=pmid, rettype="abstract", retmode="text")
        abstract = handle.read()
        handle.close()
        return abstract.strip()
    except Exception as e:
        return str(e)
```

```python
id = find_pubmed_record('10.1038/s41597-022-01513-0')
get_abstract(id)
```
