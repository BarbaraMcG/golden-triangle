```python
import multiprocessing
```

```python
def worker_func(arg1, arg2):
    try:
        result = int(arg1) + int(arg2)
        return arg1, arg2, result
    except Exception as e:
        raise Exception(arg1, arg2, e)
```

```python
results = []
def process_result(result):
    results.append(result)

def error_handler(error):
    results.append(str(error))

pool = multiprocessing.Pool(multiprocessing.cpu_count())

tasks = [('1', '2'), ('3', '4'), ('hi', '6')]

for args in tasks:
    # Non-blocking task submission: pool.apply_async() submits tasks to the process pool,
    #   allowing them to be executed in parallel without blocking the main thread.
    pool.apply_async(worker_func, args=args, callback=process_result, error_callback=error_handler)

pool.close()
pool.join()
```

```python
print(results)
```
